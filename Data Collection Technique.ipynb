Using Beautiful Soup for Data Collection

```python
from bs4 import BeautifulSoup
```

```python
with open("Shyam.html/page1.html") as f:
    content=f.read()
```

```python
soup=BeautifulSoup(content,"html.parser")
```

```python
soup.h1
```

```python
ans=soup.find_all("h3")
ans
```

```python
for h3 in ans:
    print(h3.find("a")['title'])
```

```python
articles=soup.select("article.product_pod")
articles
```

```python
for article in articles:
    title=article.find("h3").find("a")['title']
    price=article.select_one("p.price_color").text
    print(price)
    
    
```

```python
for article in articles:
    title=article.find("h3").find("a")['title']
    price=article.select_one("p.price_color").text.split("£")[1]
    print(price)
    
    
```

```python
rating_element=article.select_one("p.star-rating")
print(rating_element['class'][1])
```

```python
items.append([title,price,rating_element])
```
